```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Data
X = np.array([[1],[2],[3],[4],[5]])
y = np.array([2, 4, 5, 4, 5])

# Train model
model = LinearRegression()
model.fit(X, y)

# Best fit parameters
m_best = model.coef_[0]
c_best = model.intercept_

print("Best fit line: y = {:.2f}x + {:.2f}".format(m_best, c_best))

# Prepare plot
plt.figure(figsize=(8,5))
plt.scatter(X, y, color='blue', label='Data points')

# Range of x for plotting lines
x_range = np.linspace(0, 6, 100)

# Number of steps to interpolate between (m,c) = (0,0) and best fit
steps = 30

for i in range(steps + 1):
    # Interpolated m and c
    m = m_best * i / steps
    c = c_best * i / steps

    # Compute predicted y
    y_pred = m * x_range + c

    # Clear previous line (for animation effect)
    plt.cla()
    plt.scatter(X, y, color='blue', label='Data points')
    plt.plot(x_range, y_pred, color='red', label=f'm={m:.2f}, c={c:.2f}')
    plt.xlim(0, 6)
    plt.ylim(0, 7)
    plt.xlabel("Hours Studied")
    plt.ylabel("Marks Scored")
    plt.legend()
    plt.title("Evolution of Best Fit Line")
    plt.pause(0.2)  # pause for animation
```
