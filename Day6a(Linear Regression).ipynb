```python
# Simple Linear Regression
#import Libraries
import numpy as py # Number calc
import pandas as pd # Using for Data table
import matplotlib.pyplot as plt # Data Visualization
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
# Create Datasets
data={'House (sqrt)':[200,300,400,500,600,700,800],
      'Price (Lakhs)':[30,35,40,45,50,55,60]
      }
df=pd.DataFrame(data)
 # 3) find out input and output variables
X=df[['House (sqrt)']]
Y=df['Price (Lakhs)']

# Train-test model
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3)

#5) Object creation of Algorithms
model=LinearRegression()
# Call fit() Function of input and output variable

model.fit(X,Y)

#Predict New value
print("Pridiction of new value is :",model.predict([[1200]]))

# 8) Calculate Metrics
MSE=mean_squared_error(Y_test,X_test)
print("Mean_squared_error value is",mean_squared_error(Y_test,X_test))
print("R2_Score value is:",r2_score(Y_test,X_test))

# 9) Data Visualization (Using Matplotlib)

plt.figure(figsize=(10,6))
plt.scatter(X,Y,color="#800020",marker='*',edgecolor="black",linewidth=0.7,alpha=0.4,label='data Points')
plt.scatter(X,Y)
plt.plot(X,model.predict(X),color='red')
plt.title('Simplr Linear Regression')
plt.xlabel('study_hour')
plt.ylabel('marks')
plt.grid()
plt.legend()
plt.show()

```
