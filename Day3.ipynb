```python
num=int(input("Enter the num value: "))
i=1
while(i<=10):
  print(f"{num}*{i}=",num*i)
  i=i+1
```

```python
fact=1
i=1
while(i<=5):
    fact*=i
    i+=1
    print("Factorial number is:",fact)
```

```python
num=int(input("Enter the num value:"))
i=2
while(i<num):
  if num%i==0:
      print("Not Prime number")
      break
  i+=1
else:
  print("Prime number")
```

```python
num=int(input("Enter the num value:"))
sum=0
i=1
while(i<=num):
  sum=sum+i
  i+=1
  print("Sum of natural numbers is:",sum)
```

```python
num=int(input("Enter the num value:"))
count=0
for i in range(1,num+1):
  if num%i==0:
    count+=1

if count==2:
  print("Prime number")
else:
  print("Not Prime number")
```

```python
def Student(*marks):
  sum=0
  for i in marks:
    print(i)
    sum=sum+i
    i+=1
    avg=len(marks)
  print("total sum is",sum)
  print("Average value is:",avg)
  print("Maximum value is:",max(marks))
  print("Minimum value is:",min(marks))

Student(1,2,3,4,5)
Student(5,10,15,25,40,55,70,95,110)
```
