```python
day=int(input("Enter the day number:"))
match day:
  case 1:
    print("Monday")
  case 2:
    print("Tuesday")
  case 3:
    print("Wednesday")
  case 4:
    print("Thursday")
  case 5:
    print("Friday")
  case 6:
    print("Saturday")
  case 1:
    print("Friday")
  case _:
    print("Invalid day")
```
